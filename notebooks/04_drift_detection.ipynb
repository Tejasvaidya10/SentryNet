# SentryNet -- Concept Drift Detection

```python
from sentrynet.config import DATA_DIR

TRAIN_SPLIT_PATH = DATA_DIR / "train_split.parquet"
TEST_SPLIT_PATH = DATA_DIR / "test_split.parquet"
DATA_AVAILABLE = TRAIN_SPLIT_PATH.exists() and TEST_SPLIT_PATH.exists()

if not DATA_AVAILABLE:
    print(f"{TRAIN_SPLIT_PATH} / {TEST_SPLIT_PATH} not found. Run 03_modeling_evaluation.ipynb first.")
```

## Load train/test splits

```python
if DATA_AVAILABLE:
    import pandas as pd

    train_df = pd.read_parquet(TRAIN_SPLIT_PATH)
    test_df = pd.read_parquet(TEST_SPLIT_PATH)
```

## Real drift across the temporal split

```python
if DATA_AVAILABLE:
    from sentrynet.drift.monitor import population_stability_index, ks_drift_test, is_drifted

    psi = population_stability_index(train_df["TransactionAmt"], test_df["TransactionAmt"])
    print("PSI on TransactionAmt:", psi, "drifted:", is_drifted(psi))
```

## Synthetic drift injection (controlled demo)

```python
if DATA_AVAILABLE:
    from sentrynet.drift.synthetic import perturb_amount
    from sentrynet.drift.monitor import population_stability_index

    perturbed = perturb_amount(test_df, multiplier=3.0)
    psi_injected = population_stability_index(train_df["TransactionAmt"], perturbed["TransactionAmt"])
    print("PSI after synthetic injection:", psi_injected)
```
